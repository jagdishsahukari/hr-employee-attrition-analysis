```python
for col in cat_vs_attrition:
    print("-----",col,"vs Attrition -------")
    print(pd.crosstab(df[col],df["attrition"]))
```

```python
plot_number = 1

plt.figure(figsize= (15,12))
for col in cat_vs_attrition:
    plt.subplot(3,3,plot_number)
    sns.countplot(data = df, x = col, hue = "attrition")
    plt.title(f"{col} vs Attrition")
    plt.xticks(rotation = 90)
    plot_number += 1
plt.tight_layout()
plt.show()
```

# stage 5: Multivariate / Deeper Business Analysis

1.OverTime + JobLevel → Attrition

```python
pd.crosstab([df['joblevel'], df['overtime']],df['attrition'],normalize='index') * 100
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='joblevel',
    y='employeenumber',
    hue='overtime'
)
```

```python
joblevel_overtime_attrition = (
    df.groupby(['joblevel', 'overtime'])['attrition'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index(name='AttritionRate'))
joblevel_overtime_attrition
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=joblevel_overtime_attrition,x='joblevel',y='AttritionRate',hue='overtime')
plt.title('Attrition Rate by Job Level and OverTime')
plt.xlabel('Job Level')
plt.ylabel('Attrition Rate (%)')
plt.tight_layout()
plt.show()
```

2.JobRole + OverTime → Attrition

```python
jobrole_overtime_attrition = (df.groupby(['jobrole', 'overtime'])['attrition'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index(name='AttritionRate'))
jobrole_overtime_attrition
```

```python
plt.figure(figsize=(14, 7))
sns.barplot(
    data=jobrole_overtime_attrition,
    x='jobrole',
    y='AttritionRate',
    hue='overtime'
)
plt.title('Attrition Rate by JobRole and OverTime')
plt.xlabel('JobRole')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

3.Age + MonthlyIncome + Attrition

```python
df.groupby('attrition')[['age', 'monthlyincome']].agg(['count', 'mean', 'median', 'std'])
```

```python
group_mean = df.groupby('attrition')[['age', 'monthlyincome']].mean()

difference = (
    (group_mean.loc['No'] - group_mean.loc['Yes'])
    / group_mean.loc['No']
) * 100
difference
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='age',
    y='monthlyincome',
    hue='attrition'
)
plt.title('Age vs MonthlyIncome by Attrition')
plt.xlabel('Age')
plt.ylabel('MonthlyIncome')
plt.tight_layout()
plt.show()
```
